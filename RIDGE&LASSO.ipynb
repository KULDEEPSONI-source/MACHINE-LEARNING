# R2_SCORE

The **R² score** (or Coefficient of Determination) is a statistical metric used to evaluate the performance of a regression model. It measures how well the model predicts the target variable by quantifying the proportion of variance in the data that the model successfully "explains."

### How it Checks the Model

The R² score compares your model's predictions to a simple "baseline" model, which is just the average (mean) of all the target values.

* **The Baseline:** If you couldn't use any features to predict your outcome, your best guess for every data point would simply be the mean of all outcomes.
* **The Model's Improvement:** The R² score calculates how much better your regression line is compared to this horizontal mean line. It effectively asks: "How much of the total variation in the data is captured by my model, rather than just being left as unexplained error?"

---

### Understanding the Values

The R² score typically ranges from **0 to 1**, though it can be negative in specific cases:

* **R² = 1 (Perfect Fit):** The model predicts the target variable perfectly. The sum of squared errors is zero.
* **R² = 0 (Baseline Model):** Your model performs no better than simply predicting the mean of the target variable. It has no predictive power.
* **0 < R² < 1:** The model explains a certain percentage of the variance. For example, an R² of 0.80 means that 80% of the variation in the target variable can be explained by your input features.
* **R² < 0:** This indicates the model performs worse than the simple horizontal mean line (usually happening when the model is severely ill-fitted to the data).

---

### The Mathematical Formula

It is calculated using two key sums of squares:


$$R^2 = 1 - \frac{SSE}{SST}$$

* **SSE (Sum of Squared Errors):** The sum of the squared differences between the *actual* values and the *predicted* values. This represents the "unexplained" variance.
* **SST (Total Sum of Squares):** The sum of the squared differences between the *actual* values and the *mean* of the actual values. This represents the total variance in the data.

### Important Considerations

* **Not a Measure of "Correctness":** A high R² does not automatically mean your model is "good." It only tells you how much variance is explained. You could have a very high R² in an overfitted model that captures noise rather than actual patterns.
* **Context Matters:** What counts as a "good" score depends heavily on the field. In physical sciences, you might expect an R² near 0.99, while in social sciences, an R² of 0.30 might be considered quite informative due to the inherent complexity and randomness of human behavior.
* **Always check Residuals:** Relying solely on R² is dangerous. You should always visualize your residuals (errors) to ensure they are randomly distributed and not showing patterns, which would suggest your model is missing key information.

# alpha

In machine learning, **`alpha`** is a hyperparameter that controls the **strength of regularization**. It determines how much you want to penalize large coefficients in your model.

Think of it as a "knob" that balances two competing goals: fitting your training data perfectly versus keeping the model simple to prevent overfitting.

### How `alpha` works

The objective function for Ridge regression is:


$$\text{Minimize: } \text{Loss} + \alpha \times (\text{sum of squared coefficients})$$

* **Low `alpha` (near 0):** The penalty for large coefficients is small. The model behaves like a standard Linear Regression. It will try to fit the training data as closely as possible, which increases the risk of **overfitting** (fitting the noise in the data).
* **High `alpha`:** The penalty for large coefficients is high. The model is forced to keep its coefficients small (shrinking them toward zero). This makes the model less sensitive to individual data points, reducing the risk of **overfitting** but potentially causing **underfitting** if the value is too high.

---

### Comparison Table

| Value of `alpha` | Effect on Coefficients | Model Complexity | Risk |
| --- | --- | --- | --- |
| **0** | Original (Standard OLS) | Highest | High Overfitting |
| **Small** | Slightly reduced | High | Overfitting |
| **Large** | Heavily reduced (near 0) | Low | High Underfitting |

---

### Key Takeaways

1. **Standardization:** Because `alpha` penalizes the *size* of coefficients, Ridge and Lasso are sensitive to the scale of your features. Always scale your data (e.g., using `StandardScaler`) before using these models, so that all features are treated equally.
2. **How to choose the right value:** You don't usually guess the value of `alpha`. Instead, you use techniques like:
* **Cross-Validation:** Use `RidgeCV` or `LassoCV` in scikit-learn, which automatically tests a range of `alpha` values and selects the one that performs best on validation sets.
* **Grid Search:** Use `GridSearchCV` to test a list of potential values (e.g., `[0.01, 0.1, 1.0, 10.0]`) to find the best balance for your specific dataset.

# LASSO

```python
from sklearn.linear_model import Ridge, Lasso

# Apply Lasso Regression
lasso_model = Lasso(alpha=1.0)
lasso_model.fit(x_train, y_train)
y_pred_lasso = lasso_model.predict(x_test)
r2_model_lasso = r2_score(y_test, y_pred_lasso)
print('R2 Score of the lasso model with alpha 1.0 is: ', r2_model_lasso)
```

In the simplest terms:

`lasso_model.coef_` shows you **the weight the model assigns to each feature** to make its predictions.

The unique thing about Lasso is that **it acts like an automatic filter**:

* **Positive/Negative numbers:** These show how much—and in what direction—each feature influences the result.
* **Zero (`0.0`):** This is the "Lasso special." If a coefficient is zero, the model has **completely removed that feature** from your data, deciding it is useless for making predictions.

**In one sentence:** It tells you exactly which features matter (non-zero) and which features the model has thrown away (zero).

```python
lasso_model.coef_
```

# Identify the coefficients with bad slope results

```python
bad_features=np.where(lasso_model.coef_==0)[0]
print(bad_features)
print('Features with bad slope results', list(X.columns[bad_features]))
```

```python
# Remove these features
X_train_filtered = x_train.drop(x_train.columns[bad_features],axis=1)
X_test_filtered = x_test.drop(x_test.columns[bad_features],axis=1)

# Applying the Linear & Lasso Regression on filtered data
model_filtered_lr = LinearRegression()
model_filtered_lr.fit(X_train_filtered, y_train)

model_filtered_lasso = Lasso(alpha=0.1)
model_filtered_lasso.fit(X_train_filtered, y_train)

y_pred_lrfiltered = model_filtered_lr.predict(X_test_filtered)
r2_model_lrfiltered = r2_score(y_test, y_pred_lrfiltered)
print('R2 Score of the new Linear regression model is: ', r2_model_lrfiltered)

y_pred_lassofiltered = model_filtered_lasso.predict(X_test_filtered)
r2_model_lassofiltered = r2_score(y_test, y_pred_lassofiltered)
print('R2 Score of the new Lasso regression model is: ', r2_model_lassofiltered)
```

```python
# Identify coefficients with values closer to zero

small_coefficients = np.where((lasso_model.coef_ <0.05) & (lasso_model.coef_>-0.05))[0]
```

```python
# Removing the small coefficient features
X_train_new = x_train.drop(x_train.columns[small_coefficients],axis=1)
X_test_new = x_test.drop(x_test.columns[small_coefficients],axis=1)
```

```python
# Apply the Linear Regression
model_filtered_small = Lasso(alpha=0.1)
model_filtered_small.fit(X_train_new, y_train)

y_pred_new = model_filtered_small.predict(X_test_new)
r2_model_new = r2_score(y_test, y_pred_new)
print('R2 Score of the new Lasso regression model is: ', r2_model_new)
```

```python
from itertools import combinations

best_r2_score = 0
best_column_combination = None

# Iterate through all possible combinations of columns to drop
for r in range(1, len(small_coefficients) + 1):
    for subset in combinations(small_coefficients, r):
        # Create a new X_train with the current subset of columns dropped
        X_train_subset = x_train.drop(x_train.columns[list(subset)], axis=1)
        X_test_subset = x_test.drop(x_test.columns[list(subset)], axis=1)

        # Apply linear regression on the subset of columns
        model_subset = LinearRegression()
        model_subset.fit(X_train_subset, y_train)

        # Calculate R2 score for the subset of columns
        r2_subset = model_subset.score(X_test_subset, y_test)

        # Update the best R2 score and column combination if a higher score is achieved
        if r2_subset > best_r2_score:
            best_r2_score = r2_subset
            best_column_combination = subset

print("Best R2 score:", best_r2_score)
print("Best column combination:", list(best_column_combination))
```

This code is a **"search for the perfect subset"** script. Its goal is to find the **best possible group of features to keep** (by dropping the worst ones) to make your model as accurate as possible.

Here is the plain-English breakdown of why someone would use this:

### 1. The Core Problem: "More isn't always better"

In data science, adding more columns (features) to a model can sometimes hurt performance. This happens because:

* **Noise:** Some columns contain random data that confuses the model.
* **Redundancy:** If two columns tell the same story, keeping both can make the model "unstable."
* **Complexity:** A simpler model with fewer, high-quality features often generalizes better than a complex model with many "noisy" features.

### 2. What this code does (The "Trial and Error" process)

Imagine you have a basket of 10 "questionable" ingredients (the `small_coefficients` columns) and you are trying to make the perfect soup. You don't know which combination of ingredients makes it taste best, so you:

1. **Test:** Remove one ingredient, check the taste (R² score).
2. **Test:** Remove two, check the taste.
3. **Test:** Remove three, check the taste.
4. **Repeat:** Keep doing this until you have tried every possible combination.
5. **Declare Winner:** At the end, you look at your notes and say, *"Aha! The soup tasted best when I left out ingredients #3, #7, and #9."*

### 3. When should you actually use this?

You use this technique when:

* **You have already tried Lasso:** Lasso is a "fast" way to drop bad features, but it's a bit of a "blunt instrument." Sometimes, keeping a feature that Lasso suggested dropping (or dropping one it kept) actually results in a higher $R^2$ score.
* **Your feature set is small:** As mentioned before, you only run this if you have a handful of columns to test.
* **You want the "Absolute Best":** You are willing to trade computer time for the most optimal feature combination possible.

---

### Is this the right tool for you?

* **If you have 5–15 features to test:** This code is a great, thorough way to find the "sweet spot."
* **If you have 50+ features:** This code will likely never finish running. You should switch to **Recursive Feature Elimination (RFE)** or **Forward Selection**, which are "smarter" ways to find the best features without having to try every single combination.
